```python
import pandas as pd
import yfinance as yf
```

```python
# Download EUR/USD, EUR/GBP, EUR/JPY from Yahoo Finance (2015-2019)
pairs = {'EURUSD=X': 'EUR/USD', 'EURGBP=X': 'EUR/GBP', 'EURJPY=X': 'EUR/JPY'}
all_data = []

for ticker, name in pairs.items():
    print(f"Downloading {name}...")
    df = yf.download(ticker, start='2015-01-01', end='2019-12-31')
    df = df[['Close']].reset_index()
    df['slug'] = name
    df = df.rename(columns={'Date': 'date', 'Close': 'rate'})
    all_data.append(df)

ecb_df = pd.concat(all_data, ignore_index=True)
ecb_df.to_csv('ecb_validation_rates.csv', index=False)
print(f"\nECB validation data saved: {len(ecb_df)} rows")
```
