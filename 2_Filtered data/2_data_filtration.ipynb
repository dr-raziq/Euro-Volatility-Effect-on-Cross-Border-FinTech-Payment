```python
import pandas as pd
```

```python
# Load the full dataset
print("Loading full dataset...")
df = pd.read_csv('forex.csv')
```

```python
# Check what currency pairs are available
print("\nUnique currency pairs in dataset:")
print(df['slug'].unique())
```

```python
# Filter for EUR pairs only
eur_pairs = ['EUR/USD', 'EUR/GBP', 'EUR/JPY']
filtered_df = df[df['slug'].isin(eur_pairs)]
```

```python
# Convert date column to datetime
filtered_df['date'] = pd.to_datetime(filtered_df['date'])
```

```python
# Filter for 2015-2019 only
filtered_df = filtered_df[(filtered_df['date'] >= '2015-01-01') &
                          (filtered_df['date'] <= '2019-12-31')]
```

```python
# Sort by date
filtered_df = filtered_df.sort_values('date')
```

```python
# Save filtered dataset
filtered_df.to_csv('filtered_eur_rates.csv', index=False)
```

```python
# Print summary
print(f"\nFiltered dataset saved as 'filtered_eur_rates.csv'")
print(f"Total rows: {len(filtered_df)}")
print(f"Date range: {filtered_df['date'].min()} to {filtered_df['date'].max()}")
print(f"\nRows per currency pair:")
print(filtered_df['slug'].value_counts())
```
